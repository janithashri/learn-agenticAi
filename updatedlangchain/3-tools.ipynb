### Tools
###### Models can request to call tools that perform tasks such as fetching data froma databse , searching the web, or running code. Tools are pairings of :
###### 1. A schema, including the name of the tool , a description , and/or argument definition 
###### 2.A function or coroutine to execute

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    model="groq:llama-3.3-70b-versatile"
)
```

```python
from langchain.tools import tool

@tool
def get_weather(location: str) -> str:
    """Get the weather at alocation"""
    return f"The weather in {location} is sunny"

model_with_tools = model.bind_tools([get_weather])
```

```python
response = model_with_tools.invoke("what is the weather in tokyo?")
for tool_call in response.tool_calls:#tool_calls is a list of tool calls made by the model
    print(f"Tool call: {tool_call['name']}")
    print(f"Tool call arguments: {tool_call['args']}")


    #### tool_call is a dictionary containing the name of the tool and the arguments passed to it
    #### the tool_call['name'] is the name of the tool
    #### the tool_call['args'] is the arguments passed to the tool  
```

### Tool Execution loops

```python
# model generates tool calls
messages = [{'role': 'user', 'content': 'what is the weather in tokyo?'}]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)

# execute tool calls and get results
for tool_call in ai_msg.tool_calls:
    tool_reslt= get_weather.invoke(tool_call)
    messages.append(tool_reslt)

# pass results to model for final response
final_response = model_with_tools.invoke(messages)
print(final_response.text)
```
