### Streaming and Batch

###### most model can stream their response while it is being generated. By displaying output progressively, streaming can be a great way to improve the user experience.
###### calling stream() returns an iterator that yields the response as it is being generated.you can use a loop to print the response as it is being generated.

```python
for chunk in model.stream("write me a 200 word essay on the benefits of using langchain"):
    print(chunk.text,end="|",flush=True)
```

### Batch

###### Batching a collection of independent requests to a model can significantly improbve performance and reduce cost , as the processing can be done in parallel.

```python
responses = model.batch([
    "why are parrots so cute?",
    "what is the capital of france?",
    "what is the capital of germany?",
], config={"max_concurrency": 3} #limit the number of concurrent requests
)
for response in responses:
    print(response.content)
```
