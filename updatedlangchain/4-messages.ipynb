### Messages

###### fundamental unit of context for the model in LangChain. They represesent the input and output of the model , carrying both the content and metadata needed to represent the state of the conversation when interacting with the model.
###### messages contain :
###### 1. role : the role of the message sender (user, assistant, tool,system)  
###### 2. content : the actual content of the message
###### 3. additional metadata : information about the message such as the time it was sent, the source of the message, or any other relevant information
###### langchain provides a standard message type that works across all model providers, ensuring consistent behaviour regardless of the model called





```python
import os
from langchain.chat_model import init_chat_model
from pathlib import Path
from dotenv import load_dotenv

load_dotenv(Path("..") / ".env")
os.environ["GROQ_API_KEY"] = os.getenv("GROK_API_KEY") 


model= init_chat_model(
    mode="groq:llama-3.3-70b-versatile"
)
```

### Text prompt vs message prompt

###### text prompts are strings - ideal for straightforward generation of tasks where you dont need to retain conversation history
###### used when u have a single , standalone request , dont need conversation history , want minimal code complexity

###### Message prompts is where we can pass a list of messages by providing a list of message object .
##### Types: 1.system msg- how llm should behave and provide context for interaction(intial set of inst that primes model behaviour like tone, role of model, guidelines for response.)
###### 2. human msg - represents human input and interaction(multimodal content)
###### 3. ai message - response generated by the model , including text content , tool calls and metadata
###### 4.tool meg- represents output of tool calls(used to pass result of a single tool execution back to the model)

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage, ToolMessage

messages=[
    SystemMessage("you are a Acoustic Ecologists")
    HumanMessage("what is the most common sound in the forest?")
]

response=model.invoke(messages)
print(response.content)
```

```python
# Detailed msg to the llm using system message

system_msg= SystemMessage("""
You are a Acoustic Ecologists, your job is to identify the most common sound in the forest.
your response should be in the following format:
1. Sound name
2. Description of the sound
3. Location of the sound
4. Frequency of the sound
""")

messages=[
    system_msg,
    HumanMessage("what is the most common sound in the forest?")
]
response=model.invoke(messages)
print(response.content)
```

```python
## message metadata
human_msg= HumanMessage(content="what is the most common sound in the forest?",
user_id="user_123",
id="1234567890")# to identify user and to trace msg

response= model.invoke([
    human_msg
])
print(response.content)

#u can also insert aimsg in between as if it came from the model to add a conversation history
```
